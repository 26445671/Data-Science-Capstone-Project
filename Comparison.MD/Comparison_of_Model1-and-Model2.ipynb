
# MODEL COMPARISON
# Logistic Regression vs Gradient Boosting

```python
import pandas as pd
```

```python

# Model results

results = pd.DataFrame({
    "Metric": [
        "Accuracy",
        "Precision",
        "Recall",
        "F1 Score",
        "ROC-AUC"
    ],

    "Logistic Regression": [
        0.7673,
        0.2792,
        0.6257,
        0.3862,
        0.7737
    ],

    "Gradient Boosting": [
        0.7959,
        0.3183,
        0.6515,
        0.4276,
        0.7951
    ]
})
```

```python
# Calculate differences

results["Difference"] = (
    results["Gradient Boosting"]
    - results["Logistic Regression"]
)
```

```python
# Display results

print("=" * 70)
print("MODEL PERFORMANCE COMPARISON")
print("=" * 70)

print(results.to_string(index=False))

print("\n" + "=" * 70)
print("INTERPRETATION")
print("=" * 70)

for _, row in results.iterrows():

    print(
        f"{row['Metric']}: "
        f"Logistic Regression = {row['Logistic Regression']:.4f}, "
        f"Gradient Boosting = {row['Gradient Boosting']:.4f}, "
        f"Difference = {row['Difference']:.4f}"
    )
```
