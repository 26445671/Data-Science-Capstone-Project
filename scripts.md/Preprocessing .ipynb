```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

```python
# 1. load Dataset
df = pd.read_csv("C:/Users/ASUS/Documents/bank-full.csv", sep=";")
print(df.head)
print(df.info())
print("Shape:", df.shape)
print("Columns:", df.columns)
```

2.Data quality Check

```python
# Check duplicates
print("Duplicates:", df.duplicated().sum())
```

```python
# Check missing values
print("\nMissing values:")
print(df.isnull().sum())
```

```python
# 3. Target Encoding
df["y"] = df["y"].map({"no": 0, "yes": 1})
```

```python
# 4. Remove Leakage Variable
df = df.drop("duration", axis=1)
```

```python
 # 5. Train-Test Split

X = df.drop("y", axis=1)
y = df["y"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.3,
    random_state=42,
    stratify=y
)

print("Training size:", X_train.shape)
print("Testing size:", X_test.shape)
```

```python
# 6. One-Hot Encode Categorical Variables

X_train = pd.get_dummies(X_train, drop_first=True)
X_test = pd.get_dummies(X_test, drop_first=True)

# Align train and test sets (important after one-hot encoding)
X_train, X_test = X_train.align(X_test, join="left", axis=1, fill_value=0)
```

```python
# 7. Standardize Numerical Variables (for Logistic Regression)

num_cols = ["age", "balance", "campaign", "pdays", "previous"]

scaler = StandardScaler()
X_train_scaled = X_train.copy()
X_test_scaled = X_test.copy()

X_train_scaled[num_cols] = scaler.fit_transform(X_train[num_cols])
X_test_scaled[num_cols] = scaler.transform(X_test[num_cols])
```

```python
# 8. Class Balance Check

print("Class distribution:\n", y.value_counts(normalize=True) * 100)
```
