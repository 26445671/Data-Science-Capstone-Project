```python
# import required Libraries

import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

```python
# 1. load Dataset
df = pd.read_csv("C:/Users/ASUS/Documents/bank-full.csv", sep=";")
print(df.head)
print(df.info())
print("Shape:", df.shape)
print("Columns:", df.columns)
```

FEATURE ENGINEERING

```python
# 1. Was the customer contacted before?
df["previously_contacted"] = (df["pdays"] != -1).astype(int)
```

```python
# 2. Total number of contacts
df["total_contacts"] = df["campaign"] + df["previous"]
```

```python
# 3. Does the customer have any loan?
df["has_any_loan"] = (
    (df["housing"] == "yes") | 
    (df["loan"] == "yes")
).astype(int)
```

```python
# 4. Was the previous campaign successful?
df["previous_success"] = (df["poutcome"] == "success").astype(int)
```

```python
# 5. Age group
df["age_group"] = pd.cut(
    df["age"],
    bins=[0, 30, 45, 60, 100],
    labels=["Young", "Adult", "Middle_Age", "Senior"]
)
```

```python
# Convert target
df["y"] = df["y"].map({
    "no": 0,
    "yes": 1
})
```
